```python
# Função para obter um número válido do usuário
def get_valid_number(prompt):
    while True:
        try:
            # Solicita ao usuário um número e o converte para float
            num = float(input(prompt))
            return num
        except ValueError:
            # Captura o erro se a entrada não for um número
            print("Entrada inválida. Por favor, digite um número.")

# Solicita ao usuário o primeiro número
num1 = get_valid_number("Digite o primeiro número: ")

# Solicita ao usuário o segundo número
num2 = get_valid_number("Digite o segundo número: ")

# Realiza a operação de soma
soma = num1 + num2

# Realiza a operação de subtração
subtracao = num1 - num2

# Realiza a operação de multiplicação
multiplicacao = num1 * num2

# Realiza a operação de divisão, verificando se o divisor não é zero
if num2 != 0:
    # Formata o resultado da divisão para no máximo 4 caracteres significativos
    divisao = f"{num1 / num2:.4g}"
else:
    divisao = "Não é possível dividir por zero"

# Cria uma lista para armazenar os resultados
resultados = [
    f"Soma: {soma}",
    f"Subtração: {subtracao}",
    f"Multiplicação: {multiplicacao}",
    f"Divisão: {divisao}"
]

# Apresenta a lista de resultados
print("\nResultados:")
for resultado in resultados:
    print(resultado)
```
